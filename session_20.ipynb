```python
import pandas as pd

# Load dataset (use correct file name)
df = pd.read_csv("ford_car_dataset.csv")

# Separate features and target
X = df.drop("price", axis=1)   # independent features
y = df["price"]                # dependent feature

print("Shape of X:", X.shape)
print("Shape of y:", y.shape)

```

```python
!mv ford_car_dataset.csv ford_cleaned.csv
```

```python
# Identify categorical columns
categorical_cols = X.select_dtypes(include=["object"]).columns

# Perform one-hot encoding
X_encoded = pd.get_dummies(X, columns=categorical_cols).astype(int)

print(X_encoded.head())
```

```python
from sklearn.preprocessing import StandardScaler

# Numerical columns
num_cols = ["year", "mileage", "tax", "mpg", "engineSize"]

scaler = StandardScaler()
X_encoded[num_cols] = scaler.fit_transform(X_encoded[num_cols])

print(X_encoded[num_cols].head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.33, random_state=42
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)

print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)
```

```python
y_pred = model.predict(X_test)

print("First 10 Predictions:", y_pred[:10])
print("First 10 Actual:", y_test[:10].values)
```

```python
from sklearn.metrics import r2_score

# Calculate R² score
r2 = r2_score(y_test, y_pred)
print("R² Score:", r2)

# Interpretation
if r2 >= 0.8:
    print("The model explains most of the variance in house prices. Very good fit.")
elif r2 >= 0.5:
    print("The model explains a moderate amount of variance. Acceptable but can be improved.")
else:
    print("The model explains little variance. Performance is weak, needs better features or model.")

```

```python
import joblib

joblib.dump(model, "LR_ford_car.pkl")
joblib.dump(scaler, "scaler.pkl")
joblib.dump(X_encoded.columns.tolist(), "columns.pkl")
```
